# 7. fact_sales

```python
%sql
SELECT COUNT(*) FROM azuresalesdatabricks.gold.fact_sales;
```

```python
%sql
SELECT COUNT(*) FROM azuresalesdatabricks.gold.fact_sales WHERE customer_key IS NULL
   OR product_key IS NULL OR store_key IS NULL OR rep_key IS NULL OR date_key IS NULL;
```

```python
%sql
SELECT * FROM azuresalesdatabricks.gold.fact_sales LIMIT 10;
```

```python
%sql
SELECT 
  COUNT(*) AS total,
  COUNT(*) FILTER (WHERE customer_key IS NULL) AS null_customer,
  COUNT(*) FILTER (WHERE product_key IS NULL) AS null_product,
  COUNT(*) FILTER (WHERE store_key IS NULL) AS null_store,
  COUNT(*) FILTER (WHERE rep_key IS NULL) AS null_rep,
  COUNT(*) FILTER (WHERE date_key IS NULL) AS null_date
FROM azuresalesdatabricks.gold.fact_sales;
```

# 9. Final verification

```python
%sql
SELECT 'dim_date' AS tbl, COUNT(*) AS rows FROM azuresalesdatabricks.gold.dim_date
UNION ALL SELECT 'dim_category', COUNT(*) FROM azuresalesdatabricks.gold.dim_category
UNION ALL SELECT 'dim_customer', COUNT(*) FROM azuresalesdatabricks.gold.dim_customer
UNION ALL SELECT 'dim_product', COUNT(*) FROM azuresalesdatabricks.gold.dim_product
UNION ALL SELECT 'dim_store', COUNT(*) FROM azuresalesdatabricks.gold.dim_store
UNION ALL SELECT 'dim_sales_rep', COUNT(*) FROM azuresalesdatabricks.gold.dim_sales_rep
UNION ALL SELECT 'fact_sales', COUNT(*) FROM azuresalesdatabricks.gold.fact_sales
UNION ALL SELECT 'fact_returns', COUNT(*) FROM azuresalesdatabricks.gold.fact_returns;
```
